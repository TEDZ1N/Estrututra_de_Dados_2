```python
# Exercício 01: Ler 10 números, armazenar em um array, calcular soma e média

vetor = []          # cria uma lista vazia para armazenar os números
soma = 0            # variável para acumular a soma dos números

# Estrutura de repetição FOR: vai executar 10 vezes (i = 0 até i = 9)
for i in range(10):
    # input() exibe uma mensagem e espera o usuário digitar algo
    # int() converte o texto digitado para número inteiro
    # f-string (f"...") permite inserir variáveis dentro do texto
    numero = int(input(f"Digite o {i + 1}º número: "))

    # append() adiciona o número digitado ao final da lista 'vetor'
    vetor.append(numero)

    # soma += numero é o mesmo que soma = soma + numero
    soma += numero

# A média é calculada dividindo a soma total pela quantidade de números (10)
# O resultado é armazenado em uma variável do tipo float (número decimal)
media = soma / 10

# \n é um caractere especial que significa "nova linha" (pula uma linha)
print("\nNúmeros digitados:")

# Percorre a lista vetor e exibe cada número na tela
# end=" " faz com que os números sejam exibidos na mesma linha, separados por espaço
for numero in vetor:
    print(numero, end=" ")

# Exibe a soma total
# f-string com :.2f formata o número para 2 casas decimais
print(f"\nSoma: {soma}")
print(f"Média: {media:.2f}")
```

```python
# Exercício 02: Ler 10 números, encontrar o maior, o menor e suas posições

# Lista para armazenar os números digitados
numeros = []

# Loop para ler os 10 números
for i in range(10):
    numero = int(input(f"Digite o {i + 1}º número: "))
    numeros.append(numero)

# Inicializa as variáveis com o primeiro elemento da lista
maior = numeros[0]      # assume que o primeiro é o maior
menor = numeros[0]      # assume que o primeiro é o menor
posicao_maior = 0       # posição do maior valor
posicao_menor = 0       # posição do menor valor

# Percorre a lista usando o índice (i)
# range(len(numeros)) cria uma sequência de 0 até o tamanho da lista - 1
for i in range(len(numeros)):
    # Se o elemento atual for maior que o 'maior' armazenado...
    if numeros[i] > maior:
        maior = numeros[i]          # atualiza o maior valor
        posicao_maior = i           # guarda a posição (índice) dele

    # Se o elemento atual for menor que o 'menor' armazenado...
    if numeros[i] < menor:
        menor = numeros[i]          # atualiza o menor valor
        posicao_menor = i           # guarda a posição (índice) dele

# Exibe os resultados
# posição + 1 porque o índice começa em 0, mas o usuário vê como 1º, 2º...
print(f"\nMaior valor: {maior} (posição {posicao_maior + 1}º número)")
print(f"Menor valor: {menor} (posição {posicao_menor + 1}º número)")
```

```python
# Exercício 03: Ler 20 números, mostrar apenas os pares, contar e somar

# Lista para armazenar todos os números digitados
numeros = []

# Loop para ler 20 números
for i in range(20):
    numero = int(input(f"Digite o {i + 1}º número: "))
    numeros.append(numero)

# Lista para armazenar apenas números pares
pares = []

# Variável para acumular a soma dos números pares
soma_pares = 0

# Percorre todos os números da lista
for numero in numeros:
    # O operador % (módulo) retorna o resto da divisão
    # Se o resto da divisão por 2 for 0, o número é par
    if numero % 2 == 0:
        pares.append(numero)        # adiciona o número par à lista de pares
        soma_pares += numero        # soma o número par

# Exibe os resultados
# len(pares) retorna a quantidade de elementos na lista 'pares'
print(f"\nNúmeros pares: {pares}")
print(f"Quantidade de números pares: {len(pares)}")
print(f"Soma dos números pares: {soma_pares}")
```

```python
# Exercício 04: Ler 10 números e inverter a ordem sem usar outro array

# Lista para armazenar os números
numeros = []

# Loop para ler os 10 números
for i in range(10):
    numero = int(input(f"Digite o {i + 1}º número: "))
    numeros.append(numero)

# Exibe o vetor original
print(f"\nVetor original: {numeros}")

# Inverte o vetor usando a técnica de troca (swap)
# len(numeros) // 2 divide o tamanho por 2
# Para 10 elementos, vamos de i=0 até i=4 (5 iterações)
for i in range(len(numeros) // 2):
    # Calcula o índice do elemento simétrico
    # Exemplo: se i=0, simetrico = 9 (último elemento)
    #          se i=1, simetrico = 8 (penúltimo)
    simetrico = len(numeros) - 1 - i

    # Troca os elementos de posição usando atribuição múltipl
    #   temp = numeros[i]
    #   numeros[i] = numeros[simetrico]
    #   numeros[simetrico] = temp
    numeros[i], numeros[simetrico] = numeros[simetrico], numeros[i]

# Exibe o vetor invertido
print(f"Vetor invertido: {numeros}")
```

```python
# Exercício 05: Ler valores, exibir a matriz, somar todos e encontrar o maior

# Cria uma lista vazia para armazenar a matriz
# Uma matriz em Python é uma lista de listas
matriz = []

print("Digite os valores da matriz 3x3:")

# Loop para percorrer as linhas (i = 0, 1, 2)
for i in range(3):
    linha = []      # cria uma lista vazia para a linha atual

    # Loop para percorrer as colunas (j = 0, 1, 2)
    for j in range(3):
        # Lê o valor do elemento [i][j]
        valor = int(input(f"Elemento [{i}][{j}]: "))
        linha.append(valor)     # adiciona o valor à linha atual

    matriz.append(linha)        # adiciona a linha completa à matriz

# Exibe a matriz formatada linha por linha
print("\nMatriz:")
for linha in matriz:
    # Exibe cada linha como uma lista
    print(linha)

# Inicializa as variáveis
soma_total = 0
# Assume que o primeiro elemento é o maior
maior_valor = matriz[0][0]

# Percorre a matriz inteira
for i in range(3):          # linhas
    for j in range(3):      # colunas
        # Soma o elemento atual ao total
        soma_total += matriz[i][j]

        # Se o elemento atual for maior que o maior registrado...
        if matriz[i][j] > maior_valor:
            maior_valor = matriz[i][j]      # atualiza o maior

# Exibe os resultados
print(f"\nSoma de todos os elementos: {soma_total}")
print(f"Maior valor: {maior_valor}")
```

```python
# Exercício 06: Ler matriz, mostrar a diagonal principal e sua soma

# Cria a matriz 4x4
matriz = []

print("Digite os valores da matriz 4x4:")

# Loop para ler os valores (4 linhas)
for i in range(4):
    linha = []

    # Loop para ler os valores (4 colunas)
    for j in range(4):
        valor = int(input(f"Elemento [{i}][{j}]: "))
        linha.append(valor)

    matriz.append(linha)

# Exibe a matriz completa
print("\nMatriz:")
for linha in matriz:
    print(linha)

# Lista para armazenar os elementos da diagonal principal
diagonal = []

# Variável para somar os elementos da diagonal
soma_diagonal = 0

# Percorre a matriz (só precisamos de um loop, pois linha == coluna)
for i in range(4):
    # matriz[i][i] acessa o elemento onde linha = coluna
    diagonal.append(matriz[i][i])       # adiciona à lista
    soma_diagonal += matriz[i][i]       # soma o valor

# Exibe os resultados
print(f"\nDiagonal principal: {diagonal}")
print(f"Soma da diagonal principal: {soma_diagonal}")
```

```python
# Exercício 07: Ler notas de 4 alunos (3 provas cada) e calcular a média de cada um

# Cria a matriz de notas
# 4 linhas (alunos) x 3 colunas (provas)
notas = []

print("Digite as notas dos 4 alunos (3 provas cada):")

# Loop para percorrer os alunos (4 alunos)
for i in range(4):
    aluno = []      # lista para armazenar as notas de um aluno

    print(f"\nAluno {i + 1}:")

    # Loop para percorrer as provas (3 provas)
    for j in range(3):
        # Lê a nota da prova atual
        nota = float(input(f"  Prova {j + 1}: "))
        aluno.append(nota)      # adiciona a nota à lista do aluno

    notas.append(aluno)         # adiciona o aluno completo à matriz

# Exibe a matriz de notas formatada
print("\nMatriz de notas:")
print("         P1    P2    P3")
for i in range(4):
    print(f"Aluno {i + 1}: ", end="")
    for j in range(3):
        print(f"{notas[i][j]:6.1f}", end="")
    print()     # quebra de linha após cada aluno

# Calcula e exibe a média de cada aluno
print("\nMédias dos alunos:")
for i in range(4):
    # sum() soma todos os elementos da lista
    # Dividimos por 3 (quantidade de provas)
    media = sum(notas[i]) / 3
    print(f"Aluno {i + 1}: {media:.2f}")
```

```python
# Exercício 08: Cadastrar 5 produtos, calcular valor em estoque, encontrar o maior

# Define a classe Produto
class Produto:
    # self- objeto que está sendo criado
    def __init__(self, nome, codigo, preco, quantidade):
        self.nome = nome                # atributo: nome do produto
        self.codigo = codigo            # atributo: código do produto
        self.preco = preco              # atributo: preço unitário
        self.quantidade = quantidade    # atributo: quantidade em estoque

    # Método para calcular o valor total em estoque
    # self é passado automaticamente quando chamamos o método
    def valor_estoque(self):
        # Preço × Quantidade = valor total em estoque
        return self.preco * self.quantidade

# Lista para armazenar os produtos
produtos = []

# Loop para cadastrar 5 produtos
for i in range(5):
    print(f"\n--- Produto {i + 1} ---")

    # Lê os dados do produto
    nome = input("Nome: ")
    codigo = int(input("Código: "))
    preco = float(input("Preço: R$ "))
    quantidade = int(input("Quantidade: "))

    # Cria um novo objeto Produto com os dados lidos
    # A classe Produto é usada como um "molde" para criar objetos
    produto = Produto(nome, codigo, preco, quantidade)

    # Adiciona o produto à lista
    produtos.append(produto)

# Exibe todos os produtos cadastrados
print("\n--- PRODUTOS CADASTRADOS ---")

# Variáveis para encontrar o produto com maior valor em estoque
maior_valor = 0
produto_maior_valor = None     # None significa vazio

# Percorre a lista de produtos
for p in produtos:
    # Chama o método valor_estoque() para calcular o valor
    valor = p.valor_estoque()

    # Exibe as informações do produto
    # :.2f formata o número com 2 casas decimais
    print(f"{p.nome} (Cód: {p.codigo}) - R$ {p.preco:.2f} x {p.quantidade} = R$ {valor:.2f}")

    # Verifica se este produto tem o maior valor até agora
    if valor > maior_valor:
        maior_valor = valor
        produto_maior_valor = p

# Exibe o produto com maior valor em estoque
print(f"\nProduto com maior valor em estoque: {produto_maior_valor.nome} (R$ {maior_valor:.2f})")
```

```python
# Exercício 09: Cadastrar 5 alunos, calcular média, classificar, contar aprovados/reprovados

# Define a classe Aluno
class Aluno:
    # Construtor da classe
    def __init__(self, nome, idade, nota1, nota2, nota3):
        self.nome = nome            # nome do aluno
        self.idade = idade          # idade do aluno
        self.nota1 = nota1          # nota da primeira prova
        self.nota2 = nota2          # nota da segunda prova
        self.nota3 = nota3          # nota da terceira prova

    # Método para calcular a média das 3 notas
    def media(self):
        # Soma as 3 notas e divide por 3
        return (self.nota1 + self.nota2 + self.nota3) / 3

    # Método para verificar a situação do aluno
    def situacao(self):
        # Se a média for maior ou igual a 7.0, está aprovado
        if self.media() >= 7.0:
            return "Aprovado"
        else:
            return "Reprovado"

# Lista para armazenar os alunos
alunos = []

# Loop para cadastrar 5 alunos
for i in range(5):
    print(f"\n--- Aluno {i + 1} ---")

    # Lê os dados do aluno
    nome = input("Nome: ")
    idade = int(input("Idade: "))
    nota1 = float(input("Nota 1: "))
    nota2 = float(input("Nota 2: "))
    nota3 = float(input("Nota 3: "))

    # Cria um novo objeto Aluno
    aluno = Aluno(nome, idade, nota1, nota2, nota3)

    # Adiciona o aluno à lista
    alunos.append(aluno)

# Variáveis para as estatísticas
aprovados = 0
reprovados = 0
maior_media = 0
aluno_destaque = None

# Exibe os resultados
print("\n--- RESULTADOS ---")

# Percorre a lista de alunos
for a in alunos:
    # Calcula a média chamando o método media()
    media = a.media()

    # Verifica a situação chamando o método situacao()
    situacao = a.situacao()

    # Exibe o resultado do aluno
    print(f"{a.nome}: média {media:.2f} - {situacao}")

    # Conta aprovados e reprovados
    if situacao == "Aprovado":
        aprovados += 1
    else:
        reprovados += 1

    # Verifica se este aluno tem a maior média até agora
    if media > maior_media:
        maior_media = media
        aluno_destaque = a

# Exibe as estatísticas finais
print(f"\nAprovados: {aprovados}")
print(f"Reprovados: {reprovados}")
print(f"Aluno com maior média: {aluno_destaque.nome} ({maior_media:.2f})")
```

```python
# Exercício 10: Criar um sistema completo com menu para gerenciar 10 funcionários

# Define a classe Funcionario
class Funcionario:
    # Construtor: inicializa os atributos do funcionário
    def __init__(self, nome, idade, cargo, salario):
        self.nome = nome            # nome do funcionário
        self.idade = idade          # idade do funcionário
        self.cargo = cargo          # cargo do funcionário
        self.salario = salario      # salário do funcionário

# Lista para armazenar os funcionários
funcionarios = []

# Loop principal do programa (executa até o usuário escolher Sair)
while True:
    # Exibe o menu de opções
    # O caractere "=" cria uma linha de separação visual
    # \n no início pula uma linha antes do menu
    print("\n" + "=" * 35)
    print("    SISTEMA DE FUNCIONÁRIOS")
    print("=" * 35)
    print("1 - Cadastrar funcionários")
    print("2 - Listar funcionários")
    print("3 - Maior salário")
    print("4 - Média salarial")
    print("5 - Salários acima da média")
    print("0 - Sair")
    print("=" * 35)

    # Lê a opção escolhida pelo usuário
    # int() converte a entrada para número inteiro
    opcao = int(input("Escolha uma opção: "))

    # Opção 0: Sair do programa
    if opcao == 0:
        print("Saindo...")
        break       # break interrompe o loop while

    # Opção 1: Cadastrar funcionários
    elif opcao == 1:
        print("\n--- CADASTRAR FUNCIONÁRIOS ---")

        # Loop para cadastrar 10 funcionários
        for i in range(10):
            print(f"\nFuncionário {i + 1}:")

            # Lê os dados
            nome = input("Nome: ")
            idade = int(input("Idade: "))
            cargo = input("Cargo: ")
            salario = float(input("Salário: R$ "))

            # Cria um novo objeto Funcionario
            func = Funcionario(nome, idade, cargo, salario)

            # Adiciona à lista
            funcionarios.append(func)

        print("Cadastro concluído!")

    # Opção 2: Listar funcionários
    elif opcao == 2:
        # Verifica se a lista está vazia
        if not funcionarios:    # not funcionarios é True se a lista estiver vazia
            print("Nenhum funcionário cadastrado.")
        else:
            print("\n--- LISTA DE FUNCIONÁRIOS ---")

            # enumerate() retorna o índice e o elemento da lista
            for i, f in enumerate(funcionarios):
                print(f"{i + 1}. {f.nome} | {f.idade} anos | {f.cargo} | R$ {f.salario:.2f}")

    # Opção 3: Maior salário
    elif opcao == 3:
        if not funcionarios:
            print("Nenhum funcionário cadastrado.")
        else:
            # max() encontra o maior valor
            # key=lambda f: f.salario diz: "compare os salários"
            # lambda é uma função anônima (definida em uma linha)
            maior = max(funcionarios, key=lambda f: f.salario)
            print(f"\nFuncionário com maior salário: {maior.nome} (R$ {maior.salario:.2f})")

    # Opção 4: Média salarial
    elif opcao == 4:
        if not funcionarios:
            print("Nenhum funcionário cadastrado.")
        else:
            # Soma todos os salários usando uma list comprehension
            soma = sum(f.salario for f in funcionarios)
            # Calcula a média
            media = soma / len(funcionarios)
            print(f"\nMédia salarial: R$ {media:.2f}")

    # Opção 5: Salários acima da média
    elif opcao == 5:
        if not funcionarios:
            print("Nenhum funcionário cadastrado.")
        else:
            # Calcula a média
            soma = sum(f.salario for f in funcionarios)
            media = soma / len(funcionarios)

            print("\n--- FUNCIONÁRIOS ACIMA DA MÉDIA ---")

            # List comprehension cria uma nova lista apenas com os funcionários
            # que têm salário maior que a média
            acima = [f for f in funcionarios if f.salario > media]

            # Verifica se há funcionários acima da média
            if acima:
                # Percorre e exibe cada funcionário
                for f in acima:
                    print(f"{f.nome} | {f.cargo} | R$ {f.salario:.2f}")
            else:
                print("Nenhum funcionário com salário acima da média.")

    else:
        print("Opção inválida! Tente novamente.")
```
